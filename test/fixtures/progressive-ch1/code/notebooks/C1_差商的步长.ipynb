# 差商的步长：h 越小越准吗？

每格只做一件事。输出由你自己跑出来。这里的 `central` 是一次性的演示公式；框架里的正式版本是 `project/minigrad/numeric.py` 的 `numeric_grad`。

```python
f = lambda x: x ** 3
exact = 3 * 2.0 ** 2  # f'(2) = 12
```

```python
def central(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)
```

```python
for h in [1e-1, 1e-3, 1e-5, 1e-8, 1e-12]:
    print(f"h={h:g}  error={abs(central(f, 2.0, h) - exact):.3e}")
```

观察：误差先变小、后变大。转折点大约在哪个 h？为什么 `numeric_grad` 默认取 `1e-5`？
